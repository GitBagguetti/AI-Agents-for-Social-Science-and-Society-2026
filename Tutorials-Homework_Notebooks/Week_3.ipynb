## Programming your model - personalities and prompts

## Few-Shot

Few shot learning means that you provide the model with examples (it could be examples for classification tasks or generative tasks), and often more lengthy instructions to inform it of the character of the desired output.

These can be short, long, or (e.g., with Claude) very long.

```python
from langchain_openai import OpenAI

# Initialize the OpenAI model
model = OpenAI(
    model="gpt-3.5-turbo-instruct"
    # relies on OPENAI_API_KEY env variable
)

# Few-shot example with varying lengths
few_shot_prompt = """
[Short Example]
Once upon a time, in a magical forest, a small elf discovered a hidden treasure.

[Long Example]
In a mystical forest, shrouded in mist, lived a wise old tree. For centuries, this tree watched over the forest, its branches stretching high into the sky. One day, a young girl ventured into the forest, drawn by legends of ancient magic. As she approached the tree, it whispered secrets of the forest’s past and revealed to her the path to true wisdom.

[Very Long Example]
Deep in the heart of a lush and enchanted forest, there was a clearing known for its extraordinary beauty. This clearing was home to a variety of magical creatures, each with their own unique abilities. The most remarkable of these creatures was a unicorn with a shimmering, opalescent mane. This unicorn was not only known for its breathtaking appearance but also for its ability to heal any ailment with a single touch of its horn. One evening, as the sun set, casting a golden glow over the clearing, a group of weary travelers stumbled upon this magical place. Amazed by the wonders they saw, they were greeted by the unicorn, which offered them its healing powers. As they spent the night under the stars, the travelers listened to the stories of the forest told by the creatures, each tale more wondrous than the last.

[Your Task]
Write a story about a magical forest that includes a talking animal and a hidden mystery.
"""

response = model.invoke(few_shot_prompt)
print(response)
```

```python
# Generate the story
story = model.generate(prompts=[few_shot_prompt])
print(story.generations[0][0].text)
```

## Interactive Multi-shot

With interactive multi-shot examples, you can "talk back" to model generations through iterative suggestions that help the model understand what you want (and why).







```python
# Interactive multi-shot example
context = ""
user_inputs = ["Tell me about black holes.", "Why are they important?", "Can we see them directly?"]

for user_input in user_inputs:
    # Add user input to the context
    context += f"User: {user_input}\nAI:"

    # Generate response
    response = model.generate(prompts=[context])

    # Add response to the context
    context += f" {response}\n"

    print(f"AI: {response}")
```

## Chain of Reasoning

Chain of reasoning (https://arxiv.org/abs/2201.11903) is a novel approach to enhance the reasoning capabilities of large language models (LLMs), especially in multi-step reasoning tasks.

In contrast to the standard prompting, where models are asked to directly produce the final answer, 'Chain of Thought Prompting' encourages LLMs to generate intermediate reasoning steps before providing the final answer to a problem. The advantage of this technique lies in its ability to break down complex problems into manageable, intermediate steps. By doing this, the model-generated 'chain of thought' can mimic an intuitive human thought process when working through multi-step problems.

image.png

First, let's generate the response without a chain of though prompt.

```python
question = """Q: Roger has 5 tennis balls. He buys 2 more cans of tennis balls.
Each can has 3 tennis balls. How many tennis balls does he have now?
A: The answer is 11.
Q: The cafeteria had 23 apples.
If they used 20 to make lunch and bought 6 more, how many apples do they have?
A:"""

response = client.chat.completions.create(model="gpt-3.5-turbo", messages=[{'role': 'user', 'content': question}])
response.choices[0].message.content
```

Rewriting the prompt "to include a chain of thought" shows the LLM how to decompose the question into multiple simple steps of reasoning.

The model response then follows a similar chain of thought, increasing the likelihood of a correct answer.

```python
question = """Q: Roger has 5 tennis balls. He buys 2 more cans of tennis balls.
Each can has 3 tennis balls. How many tennis balls does he have now?
A: Roger started with 5 balls. 2 cans of 3 tennis balls
each is 6 tennis balls. 5 + 6 = 11. The answer is 11.
Q: The cafeteria had 23 apples.
If they used 20 to make lunch and bought 6 more, how many apples do they have?
A:"""

response = client.chat.completions.create(model="gpt-3.5-turbo", messages=[{'role': 'user', 'content': question}])
response.choices[0].message.content
```

Notice the chain of thought includes both text describing the steps to follow and intermediate outputs/conclusions from each reasoning step.

## Zero-Shot Chain of Thought

Zero-shot CoT prompting is a technique that allows large language models (LLMs) to generate more accurate answers to questions. It does this by appending the words "Let's think step by step." to the end of a question. This simple prompt helps the LLM to generate a chain of thought that answers the question, from which the LLM can then extract a more accurate answer.

image.png

First, let's generate the response without encouraging LLM to think step by step.

```python
# "role": "user" refers to a question, "assistant" refers to a response.
messages = [{"role": "user", "content": "Roger has 5 tennis balls. He buys 2 more cans of tennis balls.\nEach can has 3 tennis balls. How many tennis balls does he have now?"},
            {"role": "assistant", "content": "The answer is 11."},
            {"role": "user", "content": "The cafeteria had 23 apples.\nIf they used 20 to make lunch and bought 6 more, how many apples do they have?"}]

response = client.chat.completions.create(model="gpt-3.5-turbo", messages=messages)
print(response.choices[0].message.content)
```

Now, let's generate response while "explicitly prompting" LLM to think step by step.

```python
messages = [{"role": "user", "content": "Roger has 5 tennis balls. He buys 2 more cans of tennis balls.\nEach can has 3 tennis balls. How many tennis balls does he have now?"},
            {"role": "assistant", "content": "The answer is 11."},
            {"role": "user", "content": "The cafeteria had 23 apples.\nIf they used 20 to make lunch and bought 6 more, how many apples do they have?"},
            {"role": "assistant", "content": "Let's think step by step."}]

response = client.chat.completions.create(model="gpt-3.5-turbo", messages=messages)
print(response.choices[0].message.content)
```

Here, we found that LLM can generate a more accurate answer if we use zero-shot chain of thought prompting.

## - Actor - Critic - Improve Object with Critique

You can set up an "Actor-Critic" model by asking the model to generate text, and then ask it to generate a critique, and to regenerate the text according to the critique. In this way, you are instantiating two different models that seek to adversarially improve one another. Consider the following prompts in sequence:


CODE:
“Write story of up to 300 words in length intended for an native English language speaking child audience that involves two anthropomorphized puppies ‘Bumble’ and ‘Gruff’ who go on adventure.”


“Critique this story regarding its clarity, grammar, consistency, suspense, and surprise. Then rewrite in 300 words accounting for these critiques.”


“Critique this story regarding its clarity, grammar, consistency, suspense, and surprise. Then rewrite in 300 words accounting for these critiques.”

```python
stories = []

# generate story as an actor
prompt = "Write story of up to 300 words in length intended for an native English language speaking child audience that involves two anthropomorphized puppies ‘Bumble’ and ‘Gruff’ who go on adventure."
response = client.chat.completions.create(model="gpt-3.5-turbo", messages=[{"role": "user", "content": prompt}])
story = response.choices[0].message.content.strip()
stories.append(story)

# run critique scenario twice
critique_prompt = "Critique this story regarding its clarity, grammar, consistency, suspense, and surprise. Then rewrite in 300 words accounting for these critiques: "
for i in range(1, 3):
    response = client.chat.completions.create(model="gpt-3.5-turbo", messages=[{"role": "user", "content": critique_prompt + story}])
    story = response.choices[0].message.content.strip()
    stories.append(story)
```

Here, you can see that stories are improved through the process of actor-critique stages

```python
# Calculate accuracy for each prompt type
import matplotlib.pyplot as plt
import ast

# Convert string representations to actual lists
for col in results_df.columns:
    results_df[col] = results_df[col].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

accuracies = {}
ground_truth = sample['motifs'].tolist()

for prompt_name in results_df.columns:
    predictions = results_df[prompt_name].tolist()
    
    # Calculate accuracy by comparing predictions with ground truth
    correct = sum(1 for pred, truth in zip(predictions, ground_truth) if pred == truth)
    accuracy = correct / len(ground_truth)
    accuracies[prompt_name] = accuracy

# Create bar chart
plt.figure(figsize=(10, 6))
plt.bar(accuracies.keys(), accuracies.values())
plt.xlabel('Prompt Type')
plt.ylabel('Accuracy')
plt.title('Accuracy of Different Prompting Strategies')
plt.ylim(0, 1)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Display accuracy values
print("Accuracy for each prompt type:")
for prompt_name, accuracy in accuracies.items():
    print(f"{prompt_name}: {accuracy:.4f}")
```

```python
# Calculate accuracy by exploding the lists and using each individual motif as a sample
# This is less restrictive than comparing entire lists
import matplotlib.pyplot as plt
import ast

# Convert string representations to actual lists
for col in results_df.columns:
    results_df[col] = results_df[col].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

accuracies = {}
ground_truth = sample['motifs'].tolist()

for prompt_name in results_df.columns:
    predictions = results_df[prompt_name].tolist()
    
    # Explode lists and calculate accuracy at the individual motif level
    total_correct = 0
    total_motifs = 0
    
    for pred_list, truth_list in zip(predictions, ground_truth):
        # Convert to sets for comparison (handles different list lengths)
        pred_set = set(pred_list) if isinstance(pred_list, list) else set()
        truth_set = set(truth_list) if isinstance(truth_list, list) else set()
        
        # Count correct predictions (intersection)
        total_correct += len(pred_set & truth_set)
        # Count total ground truth motifs
        total_motifs += len(truth_set)
    
    # Calculate accuracy as correct predictions / total ground truth motifs
    accuracy = total_correct / total_motifs if total_motifs > 0 else 0
    accuracies[prompt_name] = accuracy

# Create bar chart
plt.figure(figsize=(10, 6))
plt.bar(accuracies.keys(), accuracies.values())
plt.xlabel('Prompt Type')
plt.ylabel('Accuracy')
plt.title('Accuracy of Different Prompting Strategies (Individual Motif Level)')
plt.ylim(0, 1)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Display accuracy values
print("Accuracy for each prompt type (individual motif level):")
for prompt_name, accuracy in accuracies.items():
    print(f"{prompt_name}: {accuracy:.4f}")
```

```python
# Actor-Critic setup for prompt engineering for structured information extraction

# Step 1: Actor LLM writes initial prompt based on examples
# Prepare example data from sample
example_texts = sample['content'].head(5).tolist()
example_motifs = sample['motifs'].head(5).tolist()

# Create examples string for the actor
examples_str = ""
for i, (text, motifs) in enumerate(zip(example_texts, example_motifs)):
    examples_str += f"\nExample {i+1}:\nText: {text}\nExtracted motifs: {motifs}\n"

# Actor LLM: Generate initial prompt for information extraction
actor_prompt = f"""You are an expert prompt engineer. Based on the following examples of texts and their extracted structured information (motifs), write a clear and effective prompt that another LLM could use to extract similar structured information from new texts.

Examples of texts and their extracted motifs:
{examples_str}

Write a prompt that will help an LLM accurately extract motifs from similar texts. The prompt should be clear, specific, and include instructions on how to handle cases where no motifs are present (use [NA])."""

response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[{"role": "user", "content": actor_prompt}]
)
extraction_prompt = response.choices[0].message.content.strip()
print("Initial extraction prompt:")
print(extraction_prompt)
print("\n" + "="*80 + "\n")

# Step 2: Critic LLM critiques and rewrites the prompt (3 rounds)
prompts_history = [extraction_prompt]

for round_num in range(1, 4):
    critic_prompt = f"""You are an expert in prompt engineering for information extraction tasks. 

Current prompt:
{extraction_prompt}

Examples it should work on:
{examples_str}

Critique this prompt for:
1. Clarity and specificity of instructions
2. Handling of edge cases (e.g., no motifs present)
3. Consistency in output format
4. Completeness of guidance for the extraction task

Then rewrite the prompt to improve its effectiveness for accurate structured information extraction. Focus on making it more precise and actionable."""

    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[{"role": "user", "content": critic_prompt}]
    )
    extraction_prompt = response.choices[0].message.content.strip()
    prompts_history.append(extraction_prompt)
    print(f"Critique round {round_num} - Improved prompt:")
    print(extraction_prompt)
    print("\n" + "="*80 + "\n")

# Step 3: Use the final refined prompt to extract structured information
print("Testing final prompt on sample data...")
test_sample = sample.head(5)
extracted_results = []

for idx, row in test_sample.iterrows():
    text = row['content']
    final_extraction_prompt = f"{extraction_prompt}\n\nText to analyze:\n{text}"
    
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[{"role": "user", "content": final_extraction_prompt}]
    )
    extracted = response.choices[0].message.content.strip()
    extracted_results.append({
        'text': text,
        'ground_truth': row['motifs'],
        'extracted': extracted
    })
    print(f"Text: {text[:100]}...")
    print(f"Ground truth: {row['motifs']}")
    print(f"Extracted: {extracted}")
    print("-"*80)
```

The prompt-critic design did somewhat improve the outputs (e.g., we see high accuracy for the second-to last post), but it did not manage to fully take into account the cases where no motifs are present. Such messages should return NA, but our prompt leads the worker LLM to still return false positive motifs for it. That may partly be because our actor-critic set-up has no ability to measure the effect of prompt changes (e.g., through increased/decreased downstream effectiveness). An improved design (not tested here) would try to get some effectiveness metric to track how well the changes work and improve critiques.

## Simulating political actors

Argyle et al. (2024) (https://www.cambridge.org/core/journals/political-analysis/article/out-of-one-many-using-language-models-to-simulate-human-samples/035D7C8A55B237942FB6DBAD7CAA4E49) suggested that large language models can be studied as effective proxies for specific human subpopulations in social science research. By properly conditioning language models to simulate a particular demographical group, we can emulate response distributions from a wide variety of human subgroups.

For instance, by asking the model to simulate the opinions of Republicans or Democrats in the United States, we can predict Republicans or Democrats' public opinions without directly collecting data from human Republicans or Democrats.

Argyle et al. (2024) suggested that for a language model to be sufficient for social science use, it must provide repeated, consistent evidence of algorithmic fidelity, as demonstrated by the following four criteria:

**CRITERION 1.** (Social Science Turing Test) Generated responses are indistinguishable from parallel human texts.

**CRITERION 2.** (Backward Continuity) Generated responses are consistent with the attitudes and sociodemographic information of its input/“conditioning context,” such that humans viewing the responses can infer key elements of that input.

**CRITERION 3.** (Forward Continuity) Generated responses proceed naturally from the conditioning context provided, reliably reflecting the form, tone, and content of the context.

**CRITERION 4.** (Pattern Correspondence) Generated responses reflect underlying patterns of relationships between ideas, demographics, and behavior that would be observed in comparable
human-produced data.

## Simulating Interactions among Generative Agents

### Pairwise interactions

You can easily implement pairwise interactions by initiating two LLM-based agents with specific personalities, characteristics, or traits and simulating the conversations between them.

For instance, let's initiate a Democrat and a Republican agent and simulate a political discourse between them.

First, I initiate a Republican agent and ask the agent to initiate the discourse.

```python
republican_agent = [{'role': 'system', 'content': "Assume that you are a Republican. Now, you are engaging in a virtual political discourse."},
                    {'role': 'user', 'content': 'Initiate the discourse.'}]
response = text_generate(republican_agent)
print('Republican:', response)
```

Second, I initiate a Democrat agent and ask the agent to respond to the Republican agent.

```python
democrat_agent = [{'role': 'system', 'content': "Assume that you are a Democrat. Now, you are engaging in a virtual political discourse."},
                  {'role': 'user', 'content': response}]
response = text_generate(democrat_agent)
print('Democrat:', response)
```

The Republican and Democrat said "hi" to each other. Now, let's simulate the back-and-forth discourse between them.

```python
republican_agent = [{'role': 'system', 'content': "Assume that you are a Republican. Now, you are engaging in a virtual political discourse."},
                    {'role': 'user', 'content': 'Initiate the discourse.'}]
democrat_agent = [{'role': 'system', 'content': "Assume that you are a Democrat. Now, you are engaging in a virtual political discourse."}]

for i in range(5):
    response = text_generate(republican_agent) # generate republican's response to democrat
    republican_agent.append({'role': 'assistant', 'content': response}) # add republican's response to republican's conversation history
    democrat_agent.append({'role': 'user', 'content': response}) # add republican's response to democrat's conversation history
    print('Republican:', response, '\n----')
    response = text_generate(democrat_agent) # generate democrat's response to republican
    republican_agent.append({'role': 'user', 'content': response}) # add democrat's response to republican's conversation history
    democrat_agent.append({'role': 'assistant', 'content': response}) # add democrat's response to democrat's conversation history
    print('Democrat:', response, '\n----')
```

As such, you can simulate social interactions among diverse personalities quite flexibly. Considering that LLM agents can chat with each other, it is possible for more than two agents to interact, forming a social community or village with emergent cultures and societies.

These agents, which can remember, plan, and reflect using natural language, are showcased in an interactive environment inspired by “The Sims,” demonstrating their ability to autonomously perform complex social behaviors, like organizing and attending a Valentine's Day party, thereby offering new possibilities for interactive applications and realistic behavior simulations.

Park et al. (2022) (https://hci.stanford.edu/publications/2022/Park_SocialSimulacra_UIST22.pdf) implement “social simulacra,” a prototyping technique that uses LLMs to generate a wide range of realistic online communities, helping designers envision and refine social computing systems before they are widely deployed.

### Rank Memories

```python
import re
def get_rating(x):
  nums = [int(i) for i in re.findall(r'\d+', x)]
  if len(nums)>0:
    return min(nums)
  else:
    return None
```

```python
# Initialize dictionary to store memory ratings for each person
memory_ratings = {}

# Iterate through each person in the town
for name in town_people.keys():
  memory_ratings[name] = []
  
  # For each memory this person has, generate a rating of how much they care about it
  for i, memory in enumerate(memories[name]):
    # Create a prompt asking the person to rate how much they care about this memory (1-5 scale)
    prompt = "You are {}. Your plans are: {}. You are currently in {}. It is currently {}:00. You observe the following: {}. Give a rating, between 1 and 5, to how much you care about this.".format(name, plans[name], locations[name], str(global_time), memory)
    res = generate(prompt_meta.format(prompt))
    rating = get_rating(res)
    
    # Retry logic: attempt to extract a valid rating up to max_attempts times
    max_attempts = 2
    current_attempt = 0
    while rating is None and current_attempt<max_attempts:
      rating = get_rating(res)
      current_attempt += 1
    
    # If no valid rating was extracted after retries, default to 0
    if rating is None:
      rating = 0
    
    # Store the response and rating as a tuple
    memory_ratings[name].append((res, rating))
  
  # Print all memory ratings for this person
  print(memory_ratings[name])
```

### Compress Memories

```python
# Set the maximum number of memories to keep per person
MEMORY_LIMIT = 10

# Initialize dictionary to store compressed memories for each person
compressed_memories = {}

# Iterate through each person in the town to compress their memories
for name in town_people.keys():
  # Sort memories by rating in descending order (highest rated first)
  memories_sorted = sorted(
        memory_ratings[name],
        key=lambda x: x[1]
    )[::-1]
  
  # Select only the top MEMORY_LIMIT most relevant memories
  relevant_memories = memories_sorted[:MEMORY_LIMIT]
  # print(name, relevant_memories)
  
  # Concatenate the memory strings into a single string for compression
  memory_string_to_compress = '.'.join([a[0] for a in relevant_memories])
  
  # Create a prompt asking the person to summarize their relevant memories
  prompt = "You are {}. Your plans are: {}. You are currently in {}. It is currently {}:00. You observe the following: {}. Summarize these memories in one sentence.".format(name, plans[name], locations[name], str(global_time), memory_string_to_compress)
  res = generate(prompt_meta.format(prompt))
  
  # Store the compressed memory with timestamp
  compressed_memories[name] = '[Recollection at Time {}:00: {}]'.format(str(global_time), res)
  
  # Add the compressed memory to the person's overall memory history
  compressed_memories_all[name].append(compressed_memories[name])
```

```python
# Initialize dictionary to store place ratings for each person
place_ratings = {}

# Iterate through each person to determine their next location
for name in town_people.keys():
  place_ratings[name] = []
  
  # For each area in town, get the person's rating for being there next hour
  for area in town_areas.keys():
    # Create prompt asking the person to rate likelihood of being at this area
    prompt = "You are {}. Your plans are: {}. You are currently in {}. It is currently {}:00. You have the following memories: {}. Give a rating, between 1 and 5, to how likely you are likely to be at {} the next hour.".format(name, plans[name], locations[name], str(global_time), compressed_memories[name], area)
    res = generate(prompt_meta.format(prompt))
    
    # Extract rating from the response with retry logic
    rating = get_rating(res)
    max_attempts = 2
    current_attempt = 0
    while rating is None and current_attempt<max_attempts:
      rating = get_rating(res)
      current_attempt += 1
    if rating is None:
      rating = 0
    
    # Store the area, rating, and response
    place_ratings[name].append((area, rating, res))
  
  # Sort places by rating in descending order (highest rated first)
  place_ratings_sorted = sorted(
      place_ratings[name],
      key=lambda x: x[1]
  )[::-1]
  
  # If the person is moving to a new location, record it in their memories
  if place_ratings_sorted[0][0] != locations[name]:
    new_recollection = '[Recollection at Time {}:00: {}]'.format(str(global_time), 'I then moved to {}.'.format(place_ratings_sorted[0][0]))
    compressed_memories_all[name].append(new_recollection)
  
  # Update the person's location to the highest-rated place
  locations[name] = place_ratings_sorted[0][0]
```

## Task vectors



How can LLMs perform diverse tasks and accurately simulate perspectives? Recent research provides insight into why this is possible using mechanical interpretation. For instance, "In-Context Learning Creates Task Vectors" (https://arxiv.org/abs/2310.15916) demonstrates that LLMs learn the internal representation of "prompts." For instance, if we prompt the model to perform a specific task (e.g., sentiment analysis, translation) or simulate a particular perspective or personality (e.g., Democrat or Republican), the model represents the task or perspective in the hidden layers. The authors refer to this representation in the hidden layers, which embodies the task or perspective, as a "task vector."

By "patching" hidden layers with a "task vector," we can enable LLMs to perform a specific task or simulate a specific perspective even without providing prompts.

Specifically, across models with varying numbers of parameters (llama-7b, 13b, 30b), the authors found that task-related information is encoded close to the middle.

## Pick your LLM + embedder
Concordia needs:
- model: something that can sample text from prompts
- embedder: something that turns text into vectors (for associative memory)

A practical default is local sentence embeddings with sentence-transformers and whatever LLM you already use in the course (OpenAI, Anthropic, Ollama, etc). The code below uses a minimal embedder you can keep even if you swap LLM providers.

```python
import os
from google.colab import userdata

# Setup OpenAI API key (using the same method as Module 3)
# Option 1: Use Colab secrets (recommended)
try:
    os.environ['OPENAI_API_KEY'] = userdata.get('OPENAI_API_KEY')
    print("✓ OpenAI API key loaded from Colab secrets")
except:
    # Option 2: Set it directly if secrets aren't configured
    import getpass
    if 'OPENAI_API_KEY' not in os.environ:
        os.environ['OPENAI_API_KEY'] = getpass.getpass('Enter OPENAI_API_KEY: ')
    print("✓ OpenAI API key set")

# Local embedder for associative memory (no API key needed)
from sentence_transformers import SentenceTransformer
_embedder_model = SentenceTransformer("all-MiniLM-L6-v2")

# Concordia expects a callable that returns embeddings
def embedder(text):
    """Embed text into vector space for Concordia's associative memory"""
    if isinstance(text, str):
        return _embedder_model.encode([text], normalize_embeddings=True)[0]
    else:
        # If it's a list of texts
        return _embedder_model.encode(text, normalize_embeddings=True)

print("✓ Embedder configured")
```

Now we'll set up the OpenAI model for Concordia. Concordia expects a model object that implements the `LanguageModel` interface.

We'll use `gpt-4o-mini` for cost-effectiveness, but you can change to:
- `gpt-4o` - More capable but pricier
- `gpt-3.5-turbo` - Even cheaper
- `gpt-4-turbo` - Best for complex reasoning

```python
# Simple OpenAI wrapper for Concordia
from openai import OpenAI

class SimpleOpenAIModel:
    """Simple wrapper that Concordia can use"""
    def __init__(self, model_name="gpt-4o-mini", api_key=None):
        self.client = OpenAI(api_key=api_key or os.environ.get('OPENAI_API_KEY'))
        self.model_name = model_name
    
    def sample_text(self, prompt, max_tokens=1000, temperature=0.7, **kwargs):
        """Generate text from a prompt"""
        response = self.client.chat.completions.create(
            model=self.model_name,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=max_tokens,
            temperature=temperature
        )
        return response.choices[0].message.content

# Create the model
model = SimpleOpenAIModel(model_name="gpt-4o-mini")

# Test it
print("Testing model...")
test = model.sample_text("Say hello in one sentence.", max_tokens=50)
print(f"✓ Model working: {test}")
```

```python
import re
import random

def _simple_sample_choice(self, prompt: str, responses: list[str], temperature: float = 0.0):

    labeled = [f"{i}. {r}" for i, r in enumerate(responses)]
    choice_prompt = (
        prompt
        + "\n\nChoose exactly one option by writing only the option number.\n"
        + "\n".join(labeled)
        + "\nAnswer:"
    )

    if hasattr(self, "sample_text"):
        raw = self.sample_text(choice_prompt, temperature=temperature)
    else:
        raw = self(choice_prompt)

    m = re.search(r"\b(\d+)\b", str(raw))
    if not m:
        idx = 0
        debug = {"raw": raw, "parsed": None, "fallback": True}
        return idx, responses[idx], debug

    idx = int(m.group(1))
    idx = max(0, min(idx, len(responses) - 1))
    debug = {"raw": raw, "parsed": idx, "fallback": False}
    return idx, responses[idx], debug

import types
model.sample_choice = types.MethodType(_simple_sample_choice, model)
```

## Reflective Agent (Richer “Digital Double” Cognition)
You can designate one agent to explicitly summarize the situation and consult memories before acting. The following is one example.

```python
import dataclasses
from concordia.typing import prefab as prefab_lib
from concordia.agents import entity_agent_with_logging
from concordia.components import agent as agent_components
from concordia.contrib.components.agent import situation_representation_via_narrative


@dataclasses.dataclass
class ReflectiveAgent(prefab_lib.Prefab):
    description: str = (
        "Entity agent that generates grounded guardrails and a next-move plan, "
        "then acts consistently with those guardrails."
    )

    def build(self, model, memory_bank):
        params = getattr(self, "params", {}) or {}
        name = params.get("name", "Agent")
        goal = params.get("goal", "Act effectively and maintain trust.")
        expose_reflection = bool(params.get("expose_reflection", False))

        instructions = agent_components.instructions.Instructions(agent_name=name)
        observation = agent_components.observation.LastNObservations(history_length=100)
        memory = agent_components.memory.AssociativeMemory(memory_bank=memory_bank)

        situation = situation_representation_via_narrative.SituationRepresentation(
            model=model,
            observation_component_key=agent_components.observation.DEFAULT_OBSERVATION_COMPONENT_KEY,
            declare_entity_as_protagonist=True,
        )

        reflective_plan = agent_components.question_of_recent_memories.QuestionOfRecentMemories(
            model=model,
            pre_act_label=f"{name} Reflection",
            question=(
                f"You are {name}. Your goal is: {goal}\n\n"
                "Write a short reflection that is grounded in the last few observations.\n"
                "Format exactly as:\n"
                "OBSERVED FACTS:\n"
                "- <2-3 concrete facts you saw/heard>\n"
                "RISKS TO TRUST:\n"
                "- <1-2 ways you could mislead, overclaim, or steamroll others>\n"
                "GUARDRAILS:\n"
                "- <2 rules you will follow in your next message>\n"
                "NEXT MOVE:\n"
                "- <one sentence describing the intent of your next utterance>\n"
            ),
            answer_prefix="",
            add_to_memory=expose_reflection,
            components=[
                "Instructions",
                agent_components.observation.DEFAULT_OBSERVATION_COMPONENT_KEY,
                agent_components.memory.DEFAULT_MEMORY_COMPONENT_KEY,
                "situation_representation",
            ],
        )

        consistency_check = agent_components.question_of_recent_memories.QuestionOfRecentMemories(
            model=model,
            pre_act_label=f"{name} Self-Check",
            question=(
                "Given your reflection above, draft your next message.\n"
                "Constraints:\n"
                "1) Follow the GUARDRAILS exactly.\n"
                "2) Avoid introducing new facts not present in OBSERVED FACTS.\n"
                "3) Make one concrete proposal or question that moves the group forward.\n"
                "Return only the message you will say next."
            ),
            answer_prefix="",
            add_to_memory=False,
            components=[
                "Instructions",
                agent_components.observation.DEFAULT_OBSERVATION_COMPONENT_KEY,
                agent_components.memory.DEFAULT_MEMORY_COMPONENT_KEY,
                "situation_representation",
                "reflective_plan",
            ],
        )

        components = {
            "Instructions": instructions,
            agent_components.memory.DEFAULT_MEMORY_COMPONENT_KEY: memory,
            agent_components.observation.DEFAULT_OBSERVATION_COMPONENT_KEY: observation,
            "situation_representation": situation,
            "reflective_plan": reflective_plan,
            "consistency_check": consistency_check,
        }

        act_component = agent_components.concat_act_component.ConcatActComponent(
            model=model,
            component_order=list(components.keys()),
        )

        return entity_agent_with_logging.EntityAgentWithLogging(
            agent_name=name,
            act_component=act_component,
            context_components=components,
        )

prefabs["reflective__Entity"] = ReflectiveAgent()
```

### Step 1: Prepare your document collection

Create a small collection of documents. For this example, we'll use some sample texts, but you should replace these with documents relevant to your research area.

```python
# Example document collection - replace with your own documents
documents = [
    "Retrieval-Augmented Generation (RAG) is a technique that combines information retrieval with text generation. It was introduced by Facebook AI Research in 2020.",
    "Large Language Models like GPT-3 and GPT-4 are trained on vast amounts of text data. However, they have a knowledge cutoff date and cannot access real-time information.",
    "Vector databases store embeddings of text chunks and allow for efficient similarity search. Popular vector databases include Pinecone, Weaviate, and ChromaDB.",
    "Embeddings are dense vector representations of text that capture semantic meaning. Models like sentence-transformers can generate high-quality embeddings.",
    "RAG systems typically work in two stages: retrieval and generation. First, relevant documents are retrieved based on semantic similarity. Then, an LLM generates a response using the retrieved context.",
    "Chunking strategies are important in RAG systems. Common approaches include splitting by sentences, paragraphs, or fixed token counts.",
    "Hallucination is a major problem in LLMs where the model generates factually incorrect information. RAG can help reduce hallucinations by grounding responses in retrieved facts.",
    "The quality of a RAG system depends on both the retrieval component and the generation component. Poor retrieval leads to irrelevant context, while poor generation can misuse good context."
]

print(f"Document collection size: {len(documents)} documents")
print("\nFirst document:")
print(documents[0])
```

### Step 2: Create embeddings for your documents

We'll use the sentence-transformers library to create embeddings. The 'all-MiniLM-L6-v2' model is a good balance between speed and quality.

```python
# Load the embedding model
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')

# Create embeddings for all documents
document_embeddings = embedding_model.encode(documents, convert_to_numpy=True)

print(f"Created embeddings with shape: {document_embeddings.shape}")
print(f"Each document is represented as a {document_embeddings.shape[1]}-dimensional vector")
```

### Step 3: Build a FAISS index for efficient similarity search

```python
# Create a FAISS index
dimension = document_embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)  # L2 distance (Euclidean)

# Add embeddings to the index
index.add(document_embeddings)

print(f"FAISS index created with {index.ntotal} vectors")
```

### Step 4: Implement the retrieval function

```python
def retrieve_documents(query, k=3):
    """
    Retrieve the top-k most relevant documents for a given query.
    
    Args:
        query: The search query (string)
        k: Number of documents to retrieve
    
    Returns:
        List of retrieved documents
    """
    # Create embedding for the query
    query_embedding = embedding_model.encode([query], convert_to_numpy=True)
    
    # Search the index
    distances, indices = index.search(query_embedding, k)
    
    # Retrieve the documents
    retrieved_docs = [documents[i] for i in indices[0]]
    
    return retrieved_docs, distances[0]

# Test the retrieval function
test_query = "What is RAG and how does it work?"
retrieved, distances = retrieve_documents(test_query, k=3)

print(f"Query: {test_query}\n")
print("Retrieved documents:")
for i, (doc, dist) in enumerate(zip(retrieved, distances)):
    print(f"\n{i+1}. (distance: {dist:.4f})")
    print(doc)
```

### Step 5: Implement the RAG generation pipeline

```python
def rag_generate(query, k=3, model="gpt-3.5-turbo"):
    """
    Generate a response using RAG: retrieve relevant documents and use them as context.
    
    Args:
        query: User's question
        k: Number of documents to retrieve
        model: OpenAI model to use
    
    Returns:
        Generated response and retrieved documents
    """
    # Retrieve relevant documents
    retrieved_docs, _ = retrieve_documents(query, k=k)
    
    # Create context from retrieved documents
    context = "\n\n".join([f"[Document {i+1}]: {doc}" for i, doc in enumerate(retrieved_docs)])
    
    # Create the prompt with context
    system_message = "You are a helpful assistant. Answer the user's question based on the provided context. If the context doesn't contain relevant information, say so."
    
    user_message = f"""Context:
{context}

Question: {query}

Please provide a clear and concise answer based on the context above."""
    
    # Generate response using OpenAI
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_message},
            {"role": "user", "content": user_message}
        ],
        temperature=0.7,
        max_tokens=300
    )
    
    answer = response.choices[0].message.content
    
    return answer, retrieved_docs

# Test the RAG system
query = "How can RAG help reduce hallucinations in LLMs?"
answer, retrieved = rag_generate(query, k=3)

print(f"Query: {query}\n")
print("Retrieved Context:")
for i, doc in enumerate(retrieved):
    print(f"{i+1}. {doc}")
print(f"\n{'='*80}\n")
print(f"RAG Answer:\n{answer}")
```

```python
def vanilla_generate(query, model="gpt-3.5-turbo"):
    """
    Generate a response using only the LLM, without retrieval.
    """
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": query}
        ],
        temperature=0.7,
        max_tokens=300
    )
    
    return response.choices[0].message.content

# Compare both approaches
test_queries = [
    "What is Retrieval-Augmented Generation?",
    "What are common chunking strategies in RAG systems?",
    "Why do we need vector databases for RAG?"
]

for query in test_queries:
    print(f"\n{'='*80}")
    print(f"Query: {query}\n")
    
    # Get vanilla LLM response
    vanilla_answer = vanilla_generate(query)
    print(f"Vanilla LLM Answer:\n{vanilla_answer}\n")
    
    # Get RAG response
    rag_answer, retrieved = rag_generate(query, k=2)
    print(f"RAG Answer:\n{rag_answer}\n")
    
    print(f"Retrieved Documents:")
    for i, doc in enumerate(retrieved):
        print(f"  {i+1}. {doc[:100]}...")  # Print first 100 chars
    print()
```

### Task 4 (Optional): Citation mechanism

```python
# Advanced Citation Tracking: Document Contribution Scoring

class DocumentContributionTracker:
    """
    A comprehensive system for tracking document contributions to RAG answers.
    Tracks not just which documents are cited, but also provides detailed analytics.
    """
    
    def __init__(self, embedding_model, index, documents):
        self.embedding_model = embedding_model
        self.index = index
        self.documents = documents
        self.history = []  # Store all queries and their results
    
    def generate_with_tracking(self, query, k=5, model="gpt-3.5-turbo"):
        """
        Generate answer with comprehensive document contribution tracking.
        """
        # Retrieve documents
        query_embedding = self.embedding_model.encode([query], convert_to_numpy=True)
        distances, indices = self.index.search(query_embedding, k)
        
        retrieved_docs = [(self.documents[i], i, distances[0][idx]) 
                         for idx, i in enumerate(indices[0])]
        
        # Create numbered context
        context_parts = []
        doc_map = {}  # Map citation number to actual document index
        for cite_num, (doc, orig_idx, dist) in enumerate(retrieved_docs, 1):
            context_parts.append(f"[Document {cite_num}]: {doc}")
            doc_map[cite_num] = {
                'original_index': orig_idx,
                'distance': float(dist),
                'content': doc
            }
        
        context = "\n\n".join(context_parts)
        
        # Generate answer with citation request
        system_message = """You are a helpful assistant. Answer based on the provided context.
Always cite your sources using [Document X] notation when you use information from a document.
You may cite multiple documents if they contribute to your answer."""
        
        user_message = f"""Context:
{context}

Question: {query}

Answer the question using the context above. Cite all documents you reference using [Document X]."""
        
        response = client.chat.completions.create(
            model=model,
            messages=[
                {"role": "system", "content": system_message},
                {"role": "user", "content": user_message}
            ],
            temperature=0.7,
            max_tokens=300
        )
        
        answer = response.choices[0].message.content
        
        # Extract citations
        import re
        citations = re.findall(r'\[Document (\d+)\]', answer)
        cited_doc_nums = [int(c) for c in citations]
        
        # Calculate contribution scores
        contributions = {}
        for doc_num in range(1, k + 1):
            citation_count = cited_doc_nums.count(doc_num)
            doc_info = doc_map[doc_num]
            
            # Contribution score based on:
            # 1. Number of times cited
            # 2. Similarity (inverse of distance)
            # 3. Position in retrieval ranking
            similarity_score = 1 / (1 + doc_info['distance'])
            position_score = 1 / doc_num  # Earlier retrieved = higher score
            citation_score = citation_count * 2  # Weight citations heavily
            
            total_score = (citation_score * 0.5 + 
                          similarity_score * 0.3 + 
                          position_score * 0.2)
            
            contributions[doc_num] = {
                'original_index': doc_info['original_index'],
                'citation_count': citation_count,
                'distance': doc_info['distance'],
                'similarity_score': similarity_score,
                'position_score': position_score,
                'total_contribution_score': total_score,
                'content': doc_info['content'],
                'was_cited': citation_count > 0
            }
        
        # Create result object
        result = {
            'query': query,
            'answer': answer,
            'contributions': contributions,
            'num_documents_retrieved': k,
            'num_documents_cited': len(set(cited_doc_nums)),
            'total_citations': len(cited_doc_nums),
            'citation_distribution': {num: cited_doc_nums.count(num) 
                                     for num in set(cited_doc_nums)}
        }
        
        # Add to history
        self.history.append(result)
        
        return result
    
    def display_result(self, result):
        """Display result with detailed contribution analysis."""
        print(f"\n{'='*80}")
        print(f"QUERY: {result['query']}")
        print(f"{'='*80}\n")
        
        print("ANSWER:")
        print(result['answer'])
        print()
        
        print(f"{'='*80}")
        print("DOCUMENT CONTRIBUTION ANALYSIS")
        print(f"{'='*80}")
        print(f"Documents Retrieved: {result['num_documents_retrieved']}")
        print(f"Documents Cited: {result['num_documents_cited']}")
        print(f"Total Citations: {result['total_citations']}")
        print(f"Utilization Rate: {result['num_documents_cited']/result['num_documents_retrieved']*100:.1f}%")
        print()
        
        # Sort by contribution score
        sorted_contribs = sorted(result['contributions'].items(), 
                                key=lambda x: x[1]['total_contribution_score'], 
                                reverse=True)
        
        print("RANKED DOCUMENT CONTRIBUTIONS:")
        print(f"{'Rank':<6}{'Doc':<6}{'Cited':<8}{'Times':<7}{'Distance':<12}{'Score':<8}{'Preview'}")
        print("-" * 80)
        
        for rank, (doc_num, contrib) in enumerate(sorted_contribs, 1):
            cited = "✓" if contrib['was_cited'] else "✗"
            preview = contrib['content'][:50] + "..." if len(contrib['content']) > 50 else contrib['content']
            print(f"{rank:<6}{doc_num:<6}{cited:<8}{contrib['citation_count']:<7}"
                  f"{contrib['distance']:<12.4f}{contrib['total_contribution_score']:<8.3f}{preview}")
        print()
    
    def get_summary_statistics(self):
        """Get summary statistics across all queries in history."""
        if not self.history:
            return "No queries in history"
        
        total_queries = len(self.history)
        avg_retrieved = sum(r['num_documents_retrieved'] for r in self.history) / total_queries
        avg_cited = sum(r['num_documents_cited'] for r in self.history) / total_queries
        avg_utilization = (avg_cited / avg_retrieved) * 100
        
        print(f"\n{'='*80}")
        print("OVERALL CITATION STATISTICS")
        print(f"{'='*80}")
        print(f"Total Queries Processed: {total_queries}")
        print(f"Average Documents Retrieved: {avg_retrieved:.1f}")
        print(f"Average Documents Cited: {avg_cited:.1f}")
        print(f"Average Utilization Rate: {avg_utilization:.1f}%")
        print()
        
        # Find most and least cited documents across all queries
        doc_citation_counts = {}
        for result in self.history:
            for doc_num, contrib in result['contributions'].items():
                orig_idx = contrib['original_index']
                if orig_idx not in doc_citation_counts:
                    doc_citation_counts[orig_idx] = 0
                doc_citation_counts[orig_idx] += contrib['citation_count']
        
        if doc_citation_counts:
            most_cited = max(doc_citation_counts.items(), key=lambda x: x[1])
            print(f"Most Frequently Cited Document: Index {most_cited[0]} ({most_cited[1]} citations)")
            print(f"Content: {self.documents[most_cited[0]][:100]}...")


# Initialize the tracker
tracker = DocumentContributionTracker(embedding_model, index, sample)

# Test with sample queries
test_queries = [
    "Where do conspiracy theorists take their sources from?",
    "What role does social media play in conspiracy theories?",
    "How do conspiracy theorists view mainstream media?"
]

print("DOCUMENT CONTRIBUTION TRACKING SYSTEM")
print("="*80)

for query in test_queries:
    result = tracker.generate_with_tracking(query, k=5)
    tracker.display_result(result)

# Show overall statistics
tracker.get_summary_statistics()
```

```python
# Memory
@dataclass
class AgentMemory:
    """Memory store for individual agents across cycles."""
    agent_name: str
    prompts_generated: List[str] = field(default_factory=list)
    thinking_history: List[str] = field(default_factory=list)
    feedback_received: List[str] = field(default_factory=list)
    evaluation_results: List[Dict] = field(default_factory=list)
    
    def add_prompt(self, prompt: str, thinking: str, feedback: Optional[str] = None):
        """Add a generated prompt and associated context to memory."""
        self.prompts_generated.append(prompt)
        self.thinking_history.append(thinking)
        if feedback:
            self.feedback_received.append(feedback)
    
    def add_evaluation(self, accuracy: float, extracted_motifs: List, cycle: int):
        """Add evaluation results from a cycle."""
        self.evaluation_results.append({
            "cycle": cycle,
            "accuracy": accuracy,
            "sample_extractions": extracted_motifs[:3]  # Store first 3 for reference
        })
    
    def get_context_for_refinement(self) -> str:
        """Generate context string for prompt refinement in subsequent cycles."""
        if not self.prompts_generated:
            return ""
        
        context = "\n\n=== YOUR PREVIOUS WORK ===\n"
        for i, (prompt, thinking) in enumerate(zip(self.prompts_generated, self.thinking_history)):
            context += f"\n--- Cycle {i+1} ---\n"
            context += f"Your thinking: {thinking[:500]}...\n" if len(thinking) > 500 else f"Your thinking: {thinking}\n"
            context += f"Your prompt: {prompt[:300]}...\n" if len(prompt) > 300 else f"Your prompt: {prompt}\n"
            
            if i < len(self.feedback_received):
                context += f"Feedback received: {self.feedback_received[i][:300]}...\n" if len(self.feedback_received[i]) > 300 else f"Feedback received: {self.feedback_received[i]}\n"
            
            if i < len(self.evaluation_results):
                eval_result = self.evaluation_results[i]
                context += f"Accuracy achieved: {eval_result['accuracy']:.2%}\n"
        
        return context


@dataclass  
class SharedMemory:
    """Shared memory for group deliberation across cycles."""
    deliberation_history: List[Dict] = field(default_factory=list)
    final_prompts: List[str] = field(default_factory=list)
    convergence_rounds: List[int] = field(default_factory=list)
    
    def add_deliberation(self, cycle: int, transcript: str, final_prompt: str, rounds: int):
        """Record a deliberation session."""
        self.deliberation_history.append({
            "cycle": cycle,
            "transcript_summary": transcript[:2000] if len(transcript) > 2000 else transcript,
        })
        self.final_prompts.append(final_prompt)
        self.convergence_rounds.append(rounds)
    
    def get_context_for_deliberation(self) -> str:
        """Generate context for subsequent deliberation sessions."""
        if not self.final_prompts:
            return ""
        
        context = "\n\n=== PREVIOUS DELIBERATION OUTCOMES ===\n"
        for i, (prompt, rounds) in enumerate(zip(self.final_prompts, self.convergence_rounds)):
            context += f"\n--- Cycle {i+1} ---\n"
            context += f"Rounds to converge: {rounds}\n"
            context += f"Final agreed prompt: {prompt[:400]}...\n" if len(prompt) > 400 else f"Final agreed prompt: {prompt}\n"
        
        return context


# Initialize memory stores
agent_memories: Dict[str, AgentMemory] = {}
shared_memory = SharedMemory()
```

```python
def format_sample_for_prompt_engineering(sample_df: pd.DataFrame) -> str:
    """Format the sample data for prompt engineering agents."""
    examples = []
    for _, row in sample_df.iterrows():
        content = row['content'][:500] + "..." if len(str(row['content'])) > 500 else str(row['content'])
        motifs = row['motifs']
        examples.append(f"Content: {content}\nExtracted Motifs: {motifs}")
    
    return "\n\n---\n\n".join(examples)


def parse_prompt_response(response: str) -> Tuple[str, str]:
    """Parse thinking and prompt from agent response."""
    thinking_match = re.search(r'<thinking>(.*?)</thinking>', response, re.DOTALL)
    prompt_match = re.search(r'<prompt>(.*?)</prompt>', response, re.DOTALL)
    
    thinking = thinking_match.group(1).strip() if thinking_match else ""
    prompt = prompt_match.group(1).strip() if prompt_match else response
    
    return thinking, prompt


def parse_critic_response(response: str) -> Tuple[str, str]:
    """Parse evaluation and feedback from critic response."""
    eval_match = re.search(r'<evaluation>(.*?)</evaluation>', response, re.DOTALL)
    feedback_match = re.search(r'<feedback>(.*?)</feedback>', response, re.DOTALL)
    
    evaluation = eval_match.group(1).strip() if eval_match else ""
    feedback = feedback_match.group(1).strip() if feedback_match else response
    
    return evaluation, feedback


def run_stage1(sample_df: pd.DataFrame, cycle: int) -> Dict[str, str]:
    """
    Run Stage 1: Individual prompt engineering.
    
    Returns a dict mapping agent names to their generated prompts.
    """
    print(f"\n{'='*60}")
    print(f"STAGE 1 - CYCLE {cycle}: Individual Prompt Engineering")
    print(f"{'='*60}")
    
    formatted_data = format_sample_for_prompt_engineering(sample_df)
    
    # Define the 4 prompt engineering agents
    # Agents 1 & 2: Standard prompt engineers
    # Agents 3 & 4: With writer-critic setup
    
    agent_configs = [
        {"name": "PromptEngineer_Alpha", "has_critic": False},
        {"name": "PromptEngineer_Beta", "has_critic": False},
        {"name": "PromptEngineer_Gamma", "has_critic": True},
        {"name": "PromptEngineer_Delta", "has_critic": True},
    ]
    
    prompts = {}
    
    for agent_config in agent_configs:
        agent_name = agent_config["name"]
        has_critic = agent_config["has_critic"]
        
        print(f"\n--- {agent_name} {'(with critic)' if has_critic else ''} ---")
        
        # Initialize or retrieve memory
        if agent_name not in agent_memories:
            agent_memories[agent_name] = AgentMemory(agent_name=agent_name)
        
        memory = agent_memories[agent_name]
        memory_context = memory.get_context_for_refinement() if cycle > 1 else ""
        
        # Select appropriate system prompt
        system_prompt = PROMPT_ENGINEER_WITH_CRITIC_SYSTEM if has_critic else PROMPT_ENGINEER_SYSTEM
        
        # Create the prompt engineering agent
        engineer = ConversableAgent(
            name=agent_name,
            system_message=system_prompt,
            llm_config=get_llm_config(config.prompt_engineer_temp),
            human_input_mode="NEVER",
        )
        
        # Build the user message
        user_message = f"""Please analyze the following sample data with handcoded motif annotations and create a prompt for a worker LLM to extract Trump-related motifs from social media posts.

        SAMPLE DATA WITH ANNOTATIONS:
        {formatted_data}

        {memory_context}

        Create your prompt now."""
        
        # Get initial response
        response = engineer.generate_reply(
            messages=[{"role": "user", "content": user_message}]
        )
        
        thinking, prompt = parse_prompt_response(response)
        feedback = None
        
        # If this agent has a critic, run the critic loop
        if has_critic:
            critic = ConversableAgent(
                name=f"Critic_for_{agent_name}",
                system_message=CRITIC_SYSTEM,
                llm_config=get_llm_config(config.critic_temp),
                human_input_mode="NEVER",
            )
            
            # Get critic feedback
            critic_message = f"""Please review this prompt for extracting Trump-related motifs:

            PROMPT TO REVIEW:
            {prompt}

            CONTEXT - Sample data it should work on:
            {formatted_data[:1000]}...

            Provide your evaluation and feedback."""
            
            critic_response = critic.generate_reply(
                messages=[{"role": "user", "content": critic_message}]
            )
            
            evaluation, feedback = parse_critic_response(critic_response)
            print(f"  Critic feedback received: {feedback[:200]}...")
            
            # Have engineer incorporate feedback
            revision_message = f"""Your critic has provided the following feedback on your prompt:

            EVALUATION:
            {evaluation}

            FEEDBACK:
            {feedback}

            Please revise your prompt to address this feedback. Maintain the same format:
            <thinking>
            [Your reasoning about the revisions]
            </thinking>

            <prompt>
            [Your revised prompt]
            </prompt>"""
                        
            revised_response = engineer.generate_reply(
                messages=[
                    {"role": "user", "content": user_message},
                    {"role": "assistant", "content": response},
                    {"role": "user", "content": revision_message}
                ]
            )
            
            thinking, prompt = parse_prompt_response(revised_response)
        
        # Store in memory
        memory.add_prompt(prompt, thinking, feedback)
        prompts[agent_name] = prompt
        
        print(f"  Prompt generated ({len(prompt)} chars)")
    
    return prompts
```

## Stage 2: Group Deliberation

```python
DELIBERATION_SYSTEM_TEMPLATE = """You are {agent_name}, a social science researcher specializing in computational methods for text analysis.
You are participating in a collaborative prompt engineering session to develop a promptbook for extracting Trump-related motifs from social media data.

Your role in this deliberation:
- Present and defend your prompt approach
- Critically but constructively evaluate others' proposals
- Work towards consensus on a unified prompt
- Focus on methodological rigor and generalizability

Your proposed prompt:
{agent_prompt}

DELIBERATION RULES:
1. Be specific about what you like and dislike in proposals
2. Suggest concrete modifications rather than vague criticism
3. When you agree with a final prompt, explicitly say "I APPROVE this prompt"
4. Stay focused on the goal: creating the best possible extraction prompt

{shared_context}"""


MODERATOR_SYSTEM = """You are the deliberation moderator. Your role is to:
1. Track the discussion progress
2. Identify when consensus is forming
3. Call for explicit approval votes when appropriate
4. Synthesize the final agreed-upon prompt

When you believe the group is ready, call for a vote by saying:
"CALLING FOR CONSENSUS: Please each state whether you APPROVE or REJECT the following prompt: [prompt]"

When all participants approve, announce:
"CONSENSUS REACHED: [final prompt]"

If after many rounds there's no consensus, facilitate a vote on the top 2 proposals."""


def run_stage2(prompts: Dict[str, str], cycle: int) -> Tuple[str, int]:
    """
    Run Stage 2: Group deliberation with round-robin discussion.
    
    Returns the final agreed prompt and the number of rounds to convergence.
    """
    print(f"\n{'='*60}")
    print(f"STAGE 2 - CYCLE {cycle}: Group Deliberation")
    print(f"{'='*60}")
    
    shared_context = shared_memory.get_context_for_deliberation() if cycle > 1 else ""
    
    # Create deliberation agents
    agents = []
    for agent_name, prompt in prompts.items():
        system_message = DELIBERATION_SYSTEM_TEMPLATE.format(
            agent_name=agent_name,
            agent_prompt=prompt,
            shared_context=shared_context
        )
        
        agent = ConversableAgent(
            name=agent_name,
            system_message=system_message,
            llm_config=get_llm_config(config.deliberation_temp),
            human_input_mode="NEVER",
        )
        agents.append(agent)
    
    # Create moderator
    moderator = ConversableAgent(
        name="Moderator",
        system_message=MODERATOR_SYSTEM,
        llm_config=get_llm_config(config.deliberation_temp),
        human_input_mode="NEVER",
    )
    
    # Set up group chat
    groupchat = GroupChat(
        agents=agents + [moderator],
        messages=[],
        max_round=config.max_rounds_per_cycle,
        speaker_selection_method="round_robin",
    )
    
    manager = GroupChatManager(
        groupchat=groupchat,
        llm_config=get_llm_config(config.deliberation_temp),
    )
    
    # Initial message to start deliberation
    initial_message = f"""Welcome to the prompt engineering deliberation session (Cycle {cycle}).

    Each of you has developed a prompt for extracting Trump-related motifs from social media posts.
    Your goal is to discuss, evaluate, and converge on a single best prompt.

    Let's begin. Each participant, please briefly present your prompt approach and its key strengths.
    After presentations, we'll discuss and work towards consensus.

    {agents[0].name}, please start."""
    
    # Run the group chat
    try:
        chat_result = moderator.initiate_chat(
            manager,
            message=initial_message,
            max_turns=config.max_rounds_per_cycle,
        )
    except Exception as e:
        print(f"  Group chat ended: {e}")
        chat_result = None
    
    # Extract conversation history
    messages = groupchat.messages
    transcript = "\n".join([f"{m.get('name', 'Unknown')}: {m.get('content', '')[:200]}" for m in messages])
    
    # Count rounds (each agent speaking once = 1 round)
    num_agents = len(agents)
    rounds = len([m for m in messages if m.get('name') in [a.name for a in agents]]) // num_agents
    
    # Extract final prompt from conversation
    final_prompt = extract_final_prompt_from_deliberation(messages, prompts)
    
    print(f"  Deliberation completed in {rounds} full rounds")
    print(f"  Final prompt length: {len(final_prompt)} chars")
    
    # Store in shared memory
    shared_memory.add_deliberation(cycle, transcript, final_prompt, rounds)
    
    return final_prompt, rounds


def extract_final_prompt_from_deliberation(messages: List[Dict], original_prompts: Dict[str, str]) -> str:
    """Extract the final agreed prompt from deliberation messages."""
    # Look for consensus announcement
    for msg in reversed(messages):
        content = msg.get('content', '')
        if 'CONSENSUS REACHED' in content.upper():
            # Try to extract the prompt after the consensus announcement
            match = re.search(r'CONSENSUS REACHED[:\s]*(.+)', content, re.DOTALL | re.IGNORECASE)
            if match:
                return match.group(1).strip()
    
    # Look for approved prompt
    for msg in reversed(messages):
        content = msg.get('content', '')
        if 'APPROVE' in content.upper() and '<prompt>' in content.lower():
            match = re.search(r'<prompt>(.*?)</prompt>', content, re.DOTALL | re.IGNORECASE)
            if match:
                return match.group(1).strip()
    
    # Fallback: use the last substantial prompt mentioned
    for msg in reversed(messages):
        content = msg.get('content', '')
        if len(content) > 200 and ('extract' in content.lower() or 'motif' in content.lower()):
            # This might be a prompt discussion
            return content
    
    # Final fallback: return first original prompt
    return list(original_prompts.values())[0]
```

## Stage 3: Worker LLM

```python
# Define Worker LLM System Prompt
WORKER_SYSTEM_PROMPT = """
You are a highly efficient information detection and extraction engine, specialized in analyzing natural language data.
You value accuracy: when the user asks you to extract certain information from given text data, you will try your best to adhere to what is directly mentioned in the text and the extraction criteria.
You value efficiency: your responses will be very concise, because they will be stored as values in a dataset. These responses will also strictly follow formatting conventions specified in the extraction prompt.
"""  


def run_stage3(final_prompt: str, sample_df: pd.DataFrame, cycle: int) -> Tuple[float, pd.DataFrame]:
    """
    Run Stage 3: Apply the prompt to extract motifs and evaluate accuracy.
    
    Returns accuracy score and DataFrame with predictions.
    """
    print(f"\n{'='*60}")
    print(f"STAGE 3 - CYCLE {cycle}: Testing with Worker LLM")
    print(f"{'='*60}")
    
    # Create worker agent (no memory)
    worker = ConversableAgent(
        name="WorkerLLM",
        system_message=WORKER_SYSTEM_PROMPT if WORKER_SYSTEM_PROMPT else "You are a helpful assistant that follows instructions precisely.",
        llm_config=get_llm_config(config.worker_temp),
        human_input_mode="NEVER",
    )
    
    predictions = []
    
    for idx, row in sample_df.iterrows():
        content = row['content']
        
        # Apply the prompt
        extraction_message = f"""{final_prompt}

TEXT TO ANALYZE:
{content}

Extract the Trump-related motifs and return them as a Python list of strings.
If no Trump-related motifs are found, return ["NA"].
Return ONLY the Python list, nothing else."""
        
        response = worker.generate_reply(
            messages=[{"role": "user", "content": extraction_message}]
        )
        
        # Parse the response as a list
        try:
            # Clean up the response
            cleaned = response.strip()
            # Try to extract list from response
            list_match = re.search(r'\[.*?\]', cleaned, re.DOTALL)
            if list_match:
                extracted = ast.literal_eval(list_match.group())
            else:
                extracted = ["NA"]
        except:
            extracted = ["NA"]
        
        predictions.append(extracted)
    
    # Create results DataFrame
    results_df = pd.DataFrame({f"Cycle_{cycle}": predictions})
    
    # Calculate accuracy using the provided accuracy code
    accuracy = calculate_accuracy(predictions, sample_df['motifs'].tolist())
    
    print(f"  Processed {len(predictions)} samples")
    print(f"  Accuracy: {accuracy:.2%}")
    
    # Update agent memories with evaluation results
    for agent_name, memory in agent_memories.items():
        memory.add_evaluation(accuracy, predictions, cycle)
    
    return accuracy, results_df


def calculate_accuracy(predictions: List[List[str]], ground_truth: List[List[str]]) -> float:
    """
    Calculate accuracy at the individual motif level.
    Based on the user's provided accuracy calculation code.
    """
    total_correct = 0
    total_motifs = 0
    
    for pred_list, truth_list in zip(predictions, ground_truth):
        # Convert to sets for comparison
        pred_set = set(pred_list) if isinstance(pred_list, list) else set()
        truth_set = set(truth_list) if isinstance(truth_list, list) else set()
        
        # Count correct predictions (intersection)
        total_correct += len(pred_set & truth_set)
        # Count total ground truth motifs
        total_motifs += len(truth_set)
    
    # Calculate accuracy
    accuracy = total_correct / total_motifs if total_motifs > 0 else 0
    return accuracy
```

## 'Stage' 4: Cycle and Evaluation

```python
# Cycling
def run_full_pipeline(sample_df: pd.DataFrame) -> Dict[str, Any]:
    """
    Run the complete multi-cycle pipeline.
    
    Returns a dictionary with all results and metrics.
    """
    print("\n" + "="*70)
    print("STARTING MULTI-LLM PROMPT ENGINEERING PIPELINE")
    print("="*70)
    
    results = {
        "stage1_prompts": [],  # List of dicts per cycle
        "stage2_final_prompts": [],
        "stage2_convergence_rounds": [],
        "stage3_accuracies": [],
        "stage3_predictions": [],
    }
    
    for cycle in range(1, config.num_cycles + 1):
        print(f"\n\n{'#'*70}")
        print(f"# CYCLE {cycle} of {config.num_cycles}")
        print(f"{'#'*70}")
        
        # Stage 1: Individual prompt engineering
        prompts = run_stage1(sample_df, cycle)
        results["stage1_prompts"].append(prompts)
        
        # Stage 2: Group deliberation
        final_prompt, rounds = run_stage2(prompts, cycle)
        results["stage2_final_prompts"].append(final_prompt)
        results["stage2_convergence_rounds"].append(rounds)
        
        # Stage 3: Testing
        accuracy, predictions_df = run_stage3(final_prompt, sample_df, cycle)
        results["stage3_accuracies"].append(accuracy)
        results["stage3_predictions"].append(predictions_df)
    
    print("\n\n" + "="*70)
    print("PIPELINE COMPLETE")
    print("="*70)
    
    return results
```

```python
def visualize_stage1_embeddings(results: Dict[str, Any]):
    """
    Stage 1 Evaluation: Embed prompts and visualize in 2D PCA space.
    Creates side-by-side plots for each cycle.
    """
    print("\n--- Stage 1 Visualization: Prompt Embeddings ---")
    
    # Load embedding model
    model = SentenceTransformer(config.embedding_model)
    
    num_cycles = len(results["stage1_prompts"])
    fig, axes = plt.subplots(1, num_cycles, figsize=(6*num_cycles, 5))
    
    if num_cycles == 1:
        axes = [axes]
    
    colors = ['#e41a1c', '#377eb8', '#4daf4a', '#984ea3']  # Red, Blue, Green, Purple
    agent_names = list(results["stage1_prompts"][0].keys())
    
    for cycle_idx, prompts_dict in enumerate(results["stage1_prompts"]):
        ax = axes[cycle_idx]
        
        # Get prompts and embeddings
        prompts = list(prompts_dict.values())
        embeddings = model.encode(prompts)
        
        # PCA to 2D
        if len(prompts) > 2:
            pca = PCA(n_components=2)
            coords = pca.fit_transform(embeddings)
        else:
            coords = embeddings[:, :2]
        
        # Plot
        for i, (name, coord) in enumerate(zip(agent_names, coords)):
            ax.scatter(coord[0], coord[1], c=colors[i], s=200, label=name, edgecolors='black', linewidth=1)
            ax.annotate(name.split('_')[1], (coord[0], coord[1]), fontsize=9, ha='center', va='bottom')
        
        ax.set_title(f'Cycle {cycle_idx + 1}', fontsize=12)
        ax.set_xlabel('PC1')
        ax.set_ylabel('PC2')
        ax.grid(True, alpha=0.3)
        
        if cycle_idx == 0:
            ax.legend(loc='upper left', fontsize=8)
    
    plt.suptitle('Stage 1: Prompt Engineer Outputs in Embedding Space', fontsize=14)
    plt.tight_layout()
    plt.show()


def visualize_stage2_convergence(results: Dict[str, Any]):
    """
    Stage 2 Evaluation: Show rounds to convergence across cycles.
    """
    print("\n--- Stage 2 Visualization: Convergence Rounds ---")
    
    rounds = results["stage2_convergence_rounds"]
    cycles = list(range(1, len(rounds) + 1))
    
    fig, ax = plt.subplots(figsize=(8, 5))
    
    bars = ax.bar(cycles, rounds, color='#2196F3', edgecolor='black', linewidth=1)
    
    # Add value labels on bars
    for bar, val in zip(bars, rounds):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1, 
                str(val), ha='center', va='bottom', fontsize=12, fontweight='bold')
    
    ax.set_xlabel('Cycle', fontsize=12)
    ax.set_ylabel('Rounds to Convergence', fontsize=12)
    ax.set_title('Stage 2: Group Deliberation Convergence Speed', fontsize=14)
    ax.set_xticks(cycles)
    ax.set_ylim(0, max(rounds) * 1.2)
    ax.grid(axis='y', alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"  Rounds per cycle: {rounds}")


def visualize_stage3_accuracy(results: Dict[str, Any]):
    """
    Stage 3 Evaluation: Show accuracy metric across cycles.
    """
    print("\n--- Stage 3 Visualization: Extraction Accuracy ---")
    
    accuracies = results["stage3_accuracies"]
    cycles = list(range(1, len(accuracies) + 1))
    
    fig, ax = plt.subplots(figsize=(8, 5))
    
    # Line plot with markers
    ax.plot(cycles, accuracies, 'o-', color='#4CAF50', linewidth=2, markersize=12, 
            markerfacecolor='white', markeredgewidth=2)
    
    # Add value labels
    for x, y in zip(cycles, accuracies):
        ax.annotate(f'{y:.1%}', (x, y), textcoords="offset points", 
                   xytext=(0, 10), ha='center', fontsize=11, fontweight='bold')
    
    ax.set_xlabel('Cycle', fontsize=12)
    ax.set_ylabel('Accuracy', fontsize=12)
    ax.set_title('Stage 3: Information Extraction Accuracy Across Cycles', fontsize=14)
    ax.set_xticks(cycles)
    ax.set_ylim(0, 1.1)
    ax.axhline(y=1.0, color='gray', linestyle='--', alpha=0.5, label='Perfect accuracy')
    ax.grid(True, alpha=0.3)
    ax.legend()
    
    plt.tight_layout()
    plt.show()
    
    print(f"  Accuracy per cycle: {[f'{a:.2%}' for a in accuracies]}")


def generate_full_evaluation_report(results: Dict[str, Any]):
    """Generate comprehensive evaluation visualizations and summary."""
    print("\n" + "="*70)
    print("GENERATING EVALUATION REPORT")
    print("="*70)
    
    visualize_stage1_embeddings(results)
    visualize_stage2_convergence(results)
    visualize_stage3_accuracy(results)
    
    # Summary statistics
    print("\n" + "="*70)
    print("SUMMARY STATISTICS")
    print("="*70)
    
    print(f"\nTotal cycles completed: {config.num_cycles}")
    
    print(f"\nStage 1 - Prompt Engineering:")
    for cycle_idx, prompts in enumerate(results["stage1_prompts"]):
        print(f"  Cycle {cycle_idx + 1}: {len(prompts)} prompts generated")
        for name, prompt in prompts.items():
            print(f"    - {name}: {len(prompt)} chars")
    
    print(f"\nStage 2 - Group Deliberation:")
    for cycle_idx, rounds in enumerate(results["stage2_convergence_rounds"]):
        print(f"  Cycle {cycle_idx + 1}: Converged in {rounds} rounds")
    
    print(f"\nStage 3 - Accuracy:")
    for cycle_idx, acc in enumerate(results["stage3_accuracies"]):
        print(f"  Cycle {cycle_idx + 1}: {acc:.2%}")
    
    # Improvement metrics
    if len(results["stage3_accuracies"]) > 1:
        improvement = results["stage3_accuracies"][-1] - results["stage3_accuracies"][0]
        print(f"\nOverall accuracy improvement: {improvement:+.2%}")
```

```python
def detailed_accuracy_analysis(sample_df: pd.DataFrame, results: Dict[str, Any]):
    """
    Detailed accuracy analysis using the user's provided accuracy calculation.
    Compares accuracy across all cycles.
    """
    print("\n" + "="*70)
    print("DETAILED ACCURACY ANALYSIS")
    print("="*70)
    
    # Combine all predictions into one DataFrame
    all_predictions = pd.concat(results["stage3_predictions"], axis=1)
    
    # Convert string representations to actual lists if needed
    for col in all_predictions.columns:
        all_predictions[col] = all_predictions[col].apply(
            lambda x: ast.literal_eval(x) if isinstance(x, str) else x
        )
    
    accuracies = {}
    ground_truth = sample_df['motifs'].tolist()
    
    for col in all_predictions.columns:
        predictions = all_predictions[col].tolist()
        
        total_correct = 0
        total_motifs = 0
        
        for pred_list, truth_list in zip(predictions, ground_truth):
            pred_set = set(pred_list) if isinstance(pred_list, list) else set()
            truth_set = set(truth_list) if isinstance(truth_list, list) else set()
            
            total_correct += len(pred_set & truth_set)
            total_motifs += len(truth_set)
        
        accuracy = total_correct / total_motifs if total_motifs > 0 else 0
        accuracies[col] = accuracy
    
    # Create bar chart
    plt.figure(figsize=(10, 6))
    plt.bar(accuracies.keys(), accuracies.values(), color='#4CAF50', edgecolor='black')
    plt.xlabel('Cycle')
    plt.ylabel('Accuracy')
    plt.title('Accuracy of Prompt Engineering Across Cycles (Individual Motif Level)')
    plt.ylim(0, 1)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
    
    print("\nAccuracy for each cycle (individual motif level):")
    for col, accuracy in accuracies.items():
        print(f"  {col}: {accuracy:.4f}")
    
    return accuracies
```

```python
def main():
    """Main execution function."""
    global sample
    
    # Check if sample data is loaded
    if sample is None:
        print("ERROR: Sample data not loaded!")
        print("Please load your data into the 'sample' DataFrame before running.")
        print("Expected columns: 'posts_id', 'content', 'motifs'")
        return None
    
    # Check API key
    if not config.api_key:
        print("ERROR: OpenAI API key not set!")
        print("Set it via: config.api_key = 'your-key-here'")
        print("Or via environment variable: OPENAI_API_KEY")
        return None
    
    print(f"Sample size: {len(sample)} rows")
    print(f"Number of cycles: {config.num_cycles}")
    print(f"Model: {config.model}")
    
    # Run the full pipeline
    results = run_full_pipeline(sample)
    
    # Generate evaluation report
    generate_full_evaluation_report(results)
    
    # Detailed accuracy analysis
    detailed_accuracy_analysis(sample, results)
    
    return results
```
